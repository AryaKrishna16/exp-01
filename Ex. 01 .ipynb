15. Adjust the brightness of the image.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('boy.jpg')

matrix_ones = np.ones(img.shape, dtype=np.float64)
matrix = matrix_ones * 50

img_brightness = cv2.add(img.astype(np.float64), matrix)
img_brightness = np.clip(img_brightness, 0, 255).astype(np.uint8)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(img_brightness, cv2.COLOR_BGR2RGB))
plt.title('Brightness Adjusted Image')
plt.axis('off')

plt.show()
```

16. Create brighter and darker images.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('boy.jpg')

matrix_ones = np.ones(img.shape, dtype=np.float64)
matrix = matrix_ones * 50

img_brighter = cv2.add(img.astype(np.float64), matrix)
img_brighter = np.clip(img_brighter, 0, 255).astype(np.uint8)

img_darker = cv2.subtract(img.astype(np.float64), matrix)
img_darker = np.clip(img_darker, 0, 255).astype(np.uint8)

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(img_darker, cv2.COLOR_BGR2RGB))
plt.title('Darker Image')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(img_brighter, cv2.COLOR_BGR2RGB))
plt.title('Brighter Image')
plt.axis('off')

plt.show()
```

17. Display the images (Original Image, Darker Image, Brighter Image).

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('boy.jpg')

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
darker_rgb = cv2.cvtColor(img_darker, cv2.COLOR_BGR2RGB)
brighter_rgb = cv2.cvtColor(img_brighter, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(img_rgb)
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(darker_rgb)
plt.title('Darker Image')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(brighter_rgb)
plt.title('Brighter Image')
plt.axis('off')

plt.tight_layout()
plt.show()
```

18. Modify the image contrast.

19. Display the images (Original, Lower Contrast, Higher Contrast).

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('boy.jpg')

matrix1 = np.ones(img.shape, dtype=np.float64) * 0.5
matrix2 = np.ones(img.shape, dtype=np.float64) * 1.5

img_lower = cv2.multiply(img.astype(np.float64), matrix1)
img_lower = np.clip(img_lower, 0, 255).astype(np.uint8)

img_higher = cv2.multiply(img.astype(np.float64), matrix2)
img_higher = np.clip(img_higher, 0, 255).astype(np.uint8)

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(img_lower, cv2.COLOR_BGR2RGB))
plt.title('Lower Contrast')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(img_higher, cv2.COLOR_BGR2RGB))
plt.title('Higher Contrast')
plt.axis('off')

plt.tight_layout()
plt.show()
```

20. Split the image (boy.jpg) into the B,G,R components & Display the channels.

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('boy.jpg')

B, G, R = cv2.split(img)

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(B, cmap='gray')
plt.title('Blue Channel')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(G, cmap='gray')
plt.title('Green Channel')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(R, cmap='gray')
plt.title('Red Channel')
plt.axis('off')

plt.tight_layout()
plt.show()
```

21. _merged the R, G, B , display along with orginal image 

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('boy.jpg')

B, G, R = cv2.split(img)

merged = cv2.merge([B, G, R])

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
merged_rgb = cv2.cvtColor(merged, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(merged_rgb)
plt.title('Merged RGB Image')
plt.axis('off')

plt.tight_layout()
plt.show()
```

22. Split the image into the H, S, V components &  Display the channels.

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('boy.jpg')

hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

H, S, V = cv2.split(hsv)

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(H, cmap='gray')
plt.title('Hue Channel')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(S, cmap='gray')
plt.title('Saturation Channel')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(V, cmap='gray')
plt.title('Value Channel')
plt.axis('off')

plt.tight_layout()
plt.show()
```

23. _merged the H, S, V, display along with orginal image 

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('boy.jpg')

hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

H, S, V = cv2.split(hsv)

merged_hsv = cv2.merge([H, S, V])

merged_bgr = cv2.cvtColor(merged_hsv, cv2.COLOR_HSV2BGR)

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
merged_rgb = cv2.cvtColor(merged_bgr, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(merged_rgb)
plt.title('Merged HSV Image')
plt.axis('off')

plt.tight_layout()
plt.show()
```
